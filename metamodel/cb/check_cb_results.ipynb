```python
import pandas as pd
import os
```

```python
# Set up paths
SCRIPT_DIR_PATH = os.getcwd()
PARENT_DIR_PATH = os.path.dirname(SCRIPT_DIR_PATH)
DATA_DIR_PATH = os.path.join(PARENT_DIR_PATH, "data")
SSP_DIR_PATH = os.path.join(DATA_DIR_PATH, "ssp")
run_id = "sisepuede_run_2025-11-12t22;19;28.194097"
RUN_DIR_PATH = os.path.join(SSP_DIR_PATH, run_id)
```

```python
# cb_df = pd.read_csv(os.path.join(RUN_DIR_PATH, f"cba_la_{run_id}_tornado.csv"))
cb_df = pd.read_csv(os.path.join(RUN_DIR_PATH, f"combined_cb_results_{run_id}.csv"))
cb_df.head()
```

```python
#print shape
cb_df.shape
```

```python
cb_df.primary_id.nunique()
```

```python
# evaluate each column null values
null_counts = cb_df.isnull().sum()
print("Null values in each column:")
print(null_counts)
```

```python
null_baseline = cb_df[cb_df['variable_value_pathway'].isnull()]
null_baseline
```

```python
null_baseline.variable.unique()  # Check unique future_ids with null baseline values
```

```python
# Check for -inf values in numeric columns
negative_inf_counts = (cb_df == -float('inf')).sum()
print("Negative infinity counts in each column:")
print(negative_inf_counts)
```

```python
# check the rows with -inf values in 'value' column
negative_inf_rows = cb_df[cb_df['value'] == -float('inf')]
print("Rows with -inf in 'value' column:")
negative_inf_rows
```

```python
# # replace -inf with 0 in 'value' column
# cb_df['value'] = cb_df['value'].replace(-float('inf'), 0)
# cb_df['difference_value'] = cb_df['difference_value'].replace(-float('inf'), 0)
```

```python
# Check for -inf values in numeric columns
negative_inf_counts = (cb_df == -float('inf')).sum()
print("Negative infinity counts in each column:")
print(negative_inf_counts)
```

```python
# replace all nans with 0
cb_df.fillna(0, inplace=True)
# Check if there are still any null values
null_counts_after_fill = cb_df.isnull().sum()
print("Null values after fillna:")
print(null_counts_after_fill)
```

```python
cb_df.primary_id.nunique()
```

```python
# Save updated DataFrame
cb_df.to_csv(os.path.join(RUN_DIR_PATH, f"combined_cb_results_{run_id}_cleaned.csv"), index=False)
print(f"Updated DataFrame saved to combined_cb_results_updated_{run_id}.csv")
```
