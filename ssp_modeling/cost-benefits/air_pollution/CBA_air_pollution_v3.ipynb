# Indoor air pollution benef

Health co-benefit from reduced household air pollution

To monetize the health benefits of reducing indoor air pollution from the use of traditional biomass stoves, we drew on estimates from the BAR-HAP model (The Benefits of Action to Reduce Household Air Pollution; WHO, 2021).
The study reports cost-effectiveness ratios ranging from 674 – 869 USD per DALY avoided for clean cooking interventions in Nepal, meaning that each avoided disability-adjusted life year (DALY) costs approximately 700 USD.

To adapt this metric to Uganda, we scaled the cost-effectiveness ratio by the relative difference in GDP per capita between Nepal (~1,450 USD) and Uganda (~1,070 USD), yielding an adjusted range of ≈ 500–640 USD per DALY avoided.
Assuming that the transition from solid fuels to clean cooking solutions avoids about 0.1 DALY per person per year, the implied annual health benefit equals 50–64 USD per person.

Given an average household size of five, this corresponds to an estimated 250–320 USD per household per year in health-related economic benefits.

This adjusted value was used in the cost–benefit analysis as the annual household health co-benefit of eliminating exposure to indoor air pollution from firewood and other solid fuels in Uganda.

```python
# --- Merge strategy attributes into ssp_data ---
# Needed to group transitions by strategy and retrieve strategy_code / future_id.
df_ssp = ssp_data.merge(
    att_primary[['primary_id', 'strategy_id', 'future_id']],
    on='primary_id', how='left'
)
df_ssp = df_ssp.merge(
    att_strategy[['strategy_id', 'strategy_code']],
    on='strategy_id', how='left'
)

# Variable maps: SISEPUEDE fuel-fraction columns -> readable transition labels.
# Used to detect which population share leaves (out_of) or enters (into) each fuel.
var_map_out = {
    'frac_scoe_heat_energy_residential_solid_biomass':             'est_out_of_biomass',
    'frac_scoe_heat_energy_residential_electricity':               'est_out_of_elec',
    'frac_scoe_heat_energy_residential_hydrocarbon_gas_liquids':   'est_out_of_gas',
}

var_map_into = {
    'frac_scoe_heat_energy_residential_solid_biomass':             'est_into_biomass',
    'frac_scoe_heat_energy_residential_electricity':               'est_into_elec',
    'frac_scoe_heat_energy_residential_hydrocarbon_gas_liquids':   'est_into_gas',
}
```

```python
# --- Helper functions to compute period-over-period fuel-fraction transitions ---

def add_out_of(df, time_col, group_cols, var_map):
    """
    out_of(t) = max(frac(t) - frac(t+1), 0)
    Captures the share of population that leaves a given fuel between t and t+1.
    Clipped at 0 so that increases in a fuel do not produce negative 'out_of' values.
    """
    df = df.sort_values(group_cols + [time_col]).copy()
    for frac_var, out_var in var_map.items():
        next_frac = df.groupby(group_cols, sort=False)[frac_var].shift(-1)
        df[out_var] = (df[frac_var] - next_frac).clip(lower=0).fillna(0)
    return df


def add_into(df, time_col, group_cols, var_map):
    """
    into(t) = max(frac(t+1) - frac(t), 0)
    Captures the share of population that enters a given fuel between t and t+1.
    Clipped at 0 so that decreases in a fuel do not produce negative 'into' values.
    """
    df = df.sort_values(group_cols + [time_col]).copy()
    for frac_var, into_var in var_map.items():
        next_frac = df.groupby(group_cols, sort=False)[frac_var].shift(-1)
        df[into_var] = (next_frac - df[frac_var]).clip(lower=0).fillna(0)
    return df
```

```python
def add_transitions(df):
    """
    Derives mutually exclusive household fuel-switching transitions per period.
    Priority order prevents double-counting:
      1. biomass -> gas  (matched first)
      2. biomass -> elec (remaining biomass leavers)
      3. gas    -> elec  (remaining electricity entrants)
    Each transition value is a fraction of total population.
    """
    df = df.copy()

    # 1. Biomass -> Gas
    df['biomass_to_gas'] = df[['est_out_of_biomass', 'est_into_gas']].min(axis=1)
    rem_out_biomass      = df['est_out_of_biomass'] - df['biomass_to_gas']
    rem_into_gas         = df['est_into_gas']       - df['biomass_to_gas']

    # 2. Biomass -> Elec (with remaining biomass leavers)
    df['biomass_to_elec'] = pd.concat([rem_out_biomass, df['est_into_elec']], axis=1).min(axis=1)
    rem_into_elec         = df['est_into_elec'] - df['biomass_to_elec']

    # 3. Gas -> Elec (with remaining electricity entrants)
    df['gas_to_elec'] = pd.concat([df['est_out_of_gas'], rem_into_elec], axis=1).min(axis=1)

    return df


# Apply: compute out_of, into, and final transitions
df_ssp = add_out_of(df_ssp,  time_col='time_period', group_cols=['strategy_id'], var_map=var_map_out)
df_ssp = add_into(df_ssp,    time_col='time_period', group_cols=['strategy_id', 'region'], var_map=var_map_into)
df_ssp = add_transitions(df_ssp)

df_ssp[['time_period', 'strategy_id', 'biomass_to_gas', 'biomass_to_elec', 'gas_to_elec']].head(10)
```

```python
# --- Per-capita cost and benefit factors by transition type ---
# Source: WHO BAR-HAP model (2021), scaled to Uganda GDP per capita.
# Four policy instruments are considered (stove subsidy, fuel subsidy, financing,
# behavioral change campaign). Because no data is available on the share of
# households reached by each instrument, factors are averaged across policies
# (unweighted). Values are in PPP 2019 USD per person.
# Costs are negative (government expenditure); benefits are positive (health gains).

df_factors_raw = pd.DataFrame([
    # --- biomass -> gas ---
    #{'transition': 'biomass_to_gas',  'policy': 'subsidy_for_stove',                    'cost_per_capita_ppp_2019': -1.349440118, 'benefit_per_capita_ppp_2019':  4.767680737},
    #{'transition': 'biomass_to_gas',  'policy': 'fuel_subsidy',                         'cost_per_capita_ppp_2019': -3.431367796, 'benefit_per_capita_ppp_2019':  5.341173225},
    {'transition': 'biomass_to_gas',  'policy': 'financing',                             'cost_per_capita_ppp_2019': -6, 'benefit_per_capita_ppp_2019': 37},
    #{'transition': 'biomass_to_gas',  'policy': 'intensive_behavioral_change_campaign',  'cost_per_capita_ppp_2019': -2.012617193, 'benefit_per_capita_ppp_2019':  6.868823172},
    # --- biomass -> elec ---
    #{'transition': 'biomass_to_elec', 'policy': 'subsidy_for_stove',                    'cost_per_capita_ppp_2019': -2.659697106, 'benefit_per_capita_ppp_2019':  9.566241834},
    #{'transition': 'biomass_to_elec', 'policy': 'fuel_subsidy',                         'cost_per_capita_ppp_2019': -6.229455845, 'benefit_per_capita_ppp_2019': 11.330111010},
    {'transition': 'biomass_to_elec', 'policy': 'financing',                             'cost_per_capita_ppp_2019': -19, 'benefit_per_capita_ppp_2019': 57},
    #{'transition': 'biomass_to_elec', 'policy': 'intensive_behavioral_change_campaign',  'cost_per_capita_ppp_2019': -3.535655711, 'benefit_per_capita_ppp_2019': 11.564970400},
    
    {'transition': 'gas_to_elec', 'policy': 'fuel_subsidy',                             'cost_per_capita_ppp_2019': -4, 'benefit_per_capita_ppp_2019': 539},
])

# Average across policies — one factor row per transition type
df_factors = (
    df_factors_raw
    .groupby('transition', as_index=False)
    .agg(
        cost_per_capita_ppp_2019=('cost_per_capita_ppp_2019',    'sum'),
        benefit_per_capita_ppp_2019=('benefit_per_capita_ppp_2019', 'sum'),
    )
)
df_factors
```

```python
# --- Monetization ---
# Only biomass_to_gas and biomass_to_elec generate indoor air pollution benefits
# (gas_to_elec does not alter indoor air quality in the same way).

transitions = ['biomass_to_gas', 'biomass_to_elec', 'gas_to_elec']

# Melt to long format: one row per (time_period, strategy_id, transition)
ssp_long = (
    df_ssp
    .melt(
        id_vars=['time_period', 'strategy_id', 'population_gnrl_total'],
        value_vars=transitions,
        var_name='transition',
        value_name='transition_frac',
    )
)

# Population affected = fraction transitioning × total population
ssp_long['population_affected'] = ssp_long['transition_frac'] * ssp_long['population_gnrl_total']

# Merge averaged per-capita factors and compute monetary values (USD PPP 2019)
cb_ap = ssp_long.merge(df_factors, on='transition', how='left')
cb_ap['total_cost']    = cb_ap['population_affected'] * cb_ap['cost_per_capita_ppp_2019']
cb_ap['total_benefit'] = cb_ap['population_affected'] * cb_ap['benefit_per_capita_ppp_2019']
cb_ap['net_benefit']   = cb_ap['total_benefit'] + cb_ap['total_cost']   # cost is negative

cb_ap.head()
```

```python
# --- Aggregate net benefit by (strategy_id, time_period) ---
# Sum net_benefit across both transitions to get a single CBA value per
# strategy/period, then attach strategy metadata for downstream use.

merged = (
    cb_ap
    .groupby(['strategy_id', 'time_period'], as_index=False)['net_benefit']
    .sum()
    .rename(columns={'net_benefit': 'value'})
)

# Exclude baseline (strategy_id == 0); benefits are relative to BAU by construction
merged = merged[merged['strategy_id'] != 0].copy()

# Re-attach strategy_code and future_id from attribute tables
merged = merged.merge(att_primary[['primary_id', 'strategy_id', 'future_id']].drop_duplicates('strategy_id'),
                      on='strategy_id', how='left')
merged = merged.merge(att_strategy[['strategy_id', 'strategy_code']], on='strategy_id', how='left')
```

```python
# Harmonize strategy name changes with a single mapping (no duplicate replacements)
strategy_name_map = {
    "PFLO:HBLE": "PFLO:HBLE",
    "PFLO:NDC_25_SCOE_ELEC": "PFLO:NDC_25_SCOE_ELEC",
    "PFLO:NDC_25_UNCONDITIONAL": "NDC_25_UNCONDITIONAL",
    "PFLO:NDC_25_UNCONDITIONAL_CE": "PFLO:NDC_25_UNCONDITIONAL_CE"
}

cb_data["strategy"] = (
    cb_data["strategy_code"].astype(str).map(strategy_name_map).fillna(cb_data["strategy_code"])
)
print(cb_data["strategy"].unique())
```

```python
# --- Strategy ID mapping based on strategy_code ---
def map_strategy_id(code: str):
    if code == "BASE":
        return 0
    if code == "PFLO:HBLE":
        return 6004
    if code == "PFLO:NDC_25":
        return 6009
    # if code == "PFLO:NDC_25_SCOE_ELEC":
    #     return 6010
    if code == "PFLO:NDC_25_UNCONDITIONAL":
        return 6011
    if code == "PFLO:NDC_25_UNCONDITIONAL_CE":
        return 6012
    return code  # keep original if not matched

cb_data["strategy_id"] = cb_data["strategy_code"].astype(str).map(map_strategy_id)
```
