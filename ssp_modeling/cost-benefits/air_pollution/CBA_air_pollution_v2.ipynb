# Indoor air pollution benef

Health co-benefit from reduced household air pollution

To monetize the health benefits of reducing indoor air pollution from the use of traditional biomass stoves, we drew on estimates from the BAR-HAP model (The Benefits of Action to Reduce Household Air Pollution; WHO, 2021).
The study reports cost-effectiveness ratios ranging from 674 – 869 USD per DALY avoided for clean cooking interventions in Nepal, meaning that each avoided disability-adjusted life year (DALY) costs approximately 700 USD.

To adapt this metric to Uganda, we scaled the cost-effectiveness ratio by the relative difference in GDP per capita between Nepal (~1,450 USD) and Uganda (~1,070 USD), yielding an adjusted range of ≈ 500–640 USD per DALY avoided.
Assuming that the transition from solid fuels to clean cooking solutions avoids about 0.1 DALY per person per year, the implied annual health benefit equals 50–64 USD per person.

Given an average household size of five, this corresponds to an estimated 250–320 USD per household per year in health-related economic benefits.

This adjusted value was used in the cost–benefit analysis as the annual household health co-benefit of eliminating exposure to indoor air pollution from firewood and other solid fuels in Uganda.

```python
var_map = {
    "frac_scoe_heat_energy_residential_solid_biomass": "est_out_of_biomass",
    "frac_scoe_heat_energy_residential_electricity":   "est_out_of_elec",
    "frac_scoe_heat_energy_residential_hydrocarbon_gas_liquids":           "est_out_of_gas",
}
```

```python
var_map_into = {
    "frac_scoe_heat_energy_residential_solid_biomass": "est_into_biomass",
    "frac_scoe_heat_energy_residential_electricity":  "est_into_elec",
    "frac_scoe_heat_energy_residential_hydrocarbon_gas_liquids":          "est_into_gas",
}
```

```python
def add_into(
    df: pd.DataFrame,
    time_col: str,
    group_cols: list[str],
    var_map: dict[str, str],
) -> pd.DataFrame:
    """
    into(t) = max(frac(t+1) - frac(t), 0)
    """
    df = df.sort_values(group_cols + [time_col]).copy()

    for frac_var, into_var in var_map.items():
        next_frac = df.groupby(group_cols, sort=False)[frac_var].shift(-1)
        df[into_var] = (next_frac - df[frac_var]).clip(lower=0).fillna(0)

    return df
```

```python
def add_out_of(
    df: pd.DataFrame,
    time_col: str,
    group_cols: list[str],
    var_map: dict[str, str],
) -> pd.DataFrame:
    """
    out_of(t) = max(frac(t) - frac(t+1), 0)
    """
    df = df.sort_values(group_cols + [time_col]).copy()

    for frac_var, out_var in var_map.items():
        next_frac = df.groupby(group_cols, sort=False)[frac_var].shift(-1)
        df[out_var] = (df[frac_var] - next_frac).clip(lower=0).fillna(0)

    return df
```

```python
df_ssp = add_out_of(
    df=ssp_data,
    time_col="time_period",                
    group_cols=["strategy_id"],   
    var_map=var_map,
)
```

```python
df_ssp = add_into(
    df=df_ssp,
    time_col="time_period",            
    group_cols=["strategy_id", "region"],
    var_map=var_map_into,
)
```

```python
def add_transitions(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # 1. Biomass -> Gas
    df["biomass_to_gas"] = df[["est_out_of_biomass", "est_into_gas"]].min(axis=1)

    # Remaining out of biomass after going to gas
    rem_out_biomass = df["est_out_of_biomass"] - df["biomass_to_gas"]

    # Remaining into gas after receiving biomass
    rem_into_gas = df["est_into_gas"] - df["biomass_to_gas"]

    # 2. Biomass -> Elec
    df["biomass_to_elec"] = pd.concat(
        [rem_out_biomass, df["est_into_elec"]],
        axis=1
    ).min(axis=1)

    # Remaining into elec after biomass
    rem_into_elec = df["est_into_elec"] - df["biomass_to_elec"]

    # 3. Gas -> Elec
    df["gas_to_elec"] = pd.concat(
        [df["est_out_of_gas"], rem_into_elec],
        axis=1
    ).min(axis=1)

    return df
```

```python
transitions = [
    "biomass_to_gas",
    "biomass_to_elec"
]

ssp_long = (
    df_ssp
    .melt(
        id_vars=["time_period",'strategy_id',"population_gnrl_total"],
        value_vars=transitions,
        var_name="transition",
        value_name="transition_frac"
    )
)
```

```python
ssp_long["population_affected"] = (
    ssp_long["transition_frac"] * ssp_long["population_gnrl_total"]
)
```

```python
cb = ssp_long.merge(df_factors, on="transition", how="left")

cb["total_cost"] = (
    cb["population_affected"] * cb["cost_per_capita_ppp_2019"]
)

cb["total_benefit"] = (
    cb["population_affected"] * cb["benefit_per_capita_ppp_2019"]
)

cb["net_benefit"] = cb["total_benefit"] - cb["total_cost"]
```

```python
cb_long = (
    cb
    .melt(
        id_vars=[
            "time_period",
            "strategy_id",
            "transition",
            "policy",
            "population_affected",
        ],
        value_vars=["total_cost", "total_benefit"],
        var_name="metric_type",
        value_name="value",
    )
)
```

```python
cb_long["metric_type"] = pd.Categorical(
    cb_long["metric_type"],
    categories=["total_benefit", "total_cost"], 
    ordered=True,
)

cb_long = cb_long.sort_values(
    by=[
        "strategy_id",
        "transition",
        "policy",
        "metric_type",   # 👈 primero benefit, luego cost
        "time_period",   # 👈 dentro de cada bloque, tiempo ascendente
    ],
    ascending=[
        True,
        True,
        True,
        True,
        True,
    ],
).reset_index(drop=True)
```
