# Estimate carrying capacity scalar to reflect baseline

```python
df_pops_projected = (
    time_periods
    .tps_to_years(
        _MODVAR_POP.get_from_dataframe(
            df_test_proj,
            fields_additional = [time_periods.field_time_period]
        )
    )
    .drop(columns = [time_periods.field_time_period])
)


# initial popualtion variables include historical numbers
df_pops_init = (
    time_periods
    .tps_to_years(
        _MODVAR_POP_INIT.get_from_dataframe(
            df_inputs_orig,
            fields_additional = [time_periods.field_time_period]
        )
    )
    .drop(columns = [time_periods.field_time_period])
)

field_check = _MODVAR_POP_INIT.build_fields(category_restrictions = "cattle_dairy", )
```

```python
ind = None
for i, row in df_pops_init.iterrows():
    if i in [0, df_pops_init.shape[0] - 1]: continue

    # 
    val = float(row[field_check])
    val_next = float(df_pops_init[field_check].iloc[i + 1])

    # the last value is filled forward; 2021 aligns with the census
    if (val == val_next) and (ind is None):
        ind = i
```

# Compare cattle populations

```python
_CAT_LVST_CATTLE_DAIRY = "cattle_dairy"
_CAT_LVST_CATTLE_NONDAIRY = "cattle_nondairy"

# get some fields
field_pop_dairy = _MODVAR_POP.build_fields(category_restrictions = _CAT_LVST_CATTLE_DAIRY)
field_pop_nondairy = _MODVAR_POP.build_fields(category_restrictions = _CAT_LVST_CATTLE_NONDAIRY)
field_pop_init_dairy = _MODVAR_POP_INIT.build_fields(category_restrictions = _CAT_LVST_CATTLE_DAIRY)
field_pop_init_nondairy = _MODVAR_POP_INIT.build_fields(category_restrictions = _CAT_LVST_CATTLE_NONDAIRY)

# set the number of years to grab the scalar for. Must be greater than `ind`
n_years = 9
n_years = max(n_years, ind)
print(f"""NOTE: need your judgement to select how many years are set for n_years. 
    Current value is {n_years} because the scalar ramps naturally back to 1
""")

vec_proj = df_pops_projected[[field_pop_dairy, field_pop_nondairy]].sum(axis = 1)
vec_empirical = df_pops_init[[field_pop_init_dairy, field_pop_init_nondairy]].sum(axis = 1)
vec_scale = (vec_empirical/vec_proj)[0:n_years]

vec_carrying_capacity = np.ones(df_inputs_orig.shape[0])
vec_carrying_capacity[0:vec_scale.shape[0]] = vec_scale

# build the output var
df_cc = _MODVAR_CC.get_from_dataframe(
    df_inputs, 
    fields_additional = [time_periods.field_year],
)

df_cc[_MODVAR_CC.fields[0]] *= vec_carrying_capacity
df_cc = (
    pd.merge(
        _DF_YEARS,
        df_cc,
        how = "left",
    )
    .interpolate()
)
```
